```python
for _, row in annot_df.iterrows():
    filename = row["XC ID"]


    begin_times = (
        row.iloc[3:]                      # columns 0,3,6,...
           .loc[lambda s: s == 1]         # keep positives
           .index.astype(float)           # column names -> times
           .tolist()
    )

    offset = row["offset"]

    audio_name = f"XC{filename} - Eurasian Woodcock - Scolopax rusticola"

    if audio_name + ".wav" in audio_files:
        sig, rate = librosa.load(audio_folder + audio_name + ".wav", sr=None, mono=True)
        sig = sig[int(rate*offset):] #### Watch out, this is onyl valid for this annotation file 
    else:
        sig, rate = librosa.load(audio_folder + audio_name + ".mp3", sr=None, mono=True)
        sig = sig[int(rate*offset):] #### Watch out, this is onyl valid for this annotation file

    extract_audio(sig, rate, begin_times, output_folder, audio_name + ".***", length)
```
