```python
output_folder= '../../data/FVA_non_Eurasian_segs/audio/seed3/0/'
```

```python
import random

for _, row in annot_df.iterrows():

    audio_name = str(row["id"])

    sig, rate = librosa.load(
        audio_folder + audio_name + ".mp3",
        sr=None,
        mono=True
    )

    # Duration in seconds
    minutes, seconds = map(int, str(row["length"]).split(":"))
    duration = minutes * 60 + seconds

    # Ignore files shorter than 3 seconds
    if duration < 3.0:
        continue

    # All possible consecutive, non-overlapping 3-second segments
    n_segments = int(duration // 3)
    begin_times = [i * 3.0 for i in range(n_segments)]

    extract_audio(
        sig,
        rate,
        begin_times,
        output_folder,
        audio_name + ".***",
        length
    )
```
